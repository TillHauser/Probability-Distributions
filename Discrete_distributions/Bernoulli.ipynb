# The Bernoulli Distribution

For $0\leq p \leq 1$ the Bernoulli distribution (w.r.t. $p$) is the discrete probability distribution of a random variable which takes the value $1$ with probability $p$ and the value $0$ with probability $1 − p$. 
If $X$ is a random variable with this distribution we write $X \sim \operatorname{Bernoulli}(p)$. 

**Remarks:**
The Bernoulli distribution with $p=0.5$ models the toss of a fair coin. For other $p$ it can be seen as a unfair coin. 

## Some Visualizations

```python
def Bernoulli(p=0.5):
    """A function that returns the probability mass 
    function of a Bernoulli random variable given 
    the parameter p."""
    if not 0 <= p <= 1:
        raise Exception("p needs to be between 0 and 1 for the bernoulli distribution.")
    def f(x: float) -> float: 
        if float(x) == 1.0: 
            return float(p)
        elif float(x)==0.0:
            return float(1-p)
        else: 
            return 0.0
    return f

# Creating a Bernoulli probability mass function
p = 0.4
f = Bernoulli(p)

# Visualization via matplotlib
from matplotlib import pyplot as plt
x_range = [0,1]
plt.bar(x_range, [f(x) for x in x_range])
plt.xlabel('x')
plt.ylabel('P(X=x)')
plt.title(rf'Bernoulli distribution ($p={p}$)')

plt.show()


plt.pie(
    [f(x) for x in x_range], 
    labels=x_range, 
    startangle=90
)
plt.title(rf'Bernoulli distribution ($p={p}$)')
plt.show()
```

## Simulation

The Bernoulli distribution is a special case of the binomial distribution and can therefore be simulated via numpy as follows.  

```python
import numpy as np

p = 0.4
x = np.random.binomial(n=1, p=p)
print(x)

# With numpy we can even get a whole sample
n = 1000
samples = np.random.binomial(n=1, p=p, size=n)
#print(samples)
print(list(samples).count(1)/n)
```

The *random* method of the *random* module gives a uniform distribution, which also allows to simulate a Bernoulli distribution as follows. 

```python
from random import random

p = 0.4
x = int(random() < p)
print(x)

# Here samples can be collected as follows
n = 1000
samples = [int(random() < p) for _ in range(n)]
#print(samples)
print(samples.count(1)/n)
```

Another possibility in the *random* module is given by the following. 

```python
from random import choices

x = choices([0, 1], weights=[1-p, p])[0]
print(x)
```
