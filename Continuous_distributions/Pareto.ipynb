# The Pareto distribution

For parameters $\kappa>0$ and $\lambda>0$ the probability density function of the *Pareto distribution* is given by 
$$
\frac{\kappa\lambda^\kappa}{x^{\kappa+1}}I_{(0,\infty)})(x).


```python
from matplotlib import pyplot as plt
import numpy as np
from math import sqrt

k = 1.5 # kappa
l = 1.5 # lambda

step = 0.05

def f(x):
    if x <= l:
        return 0
    return ( k * l**k ) / ( x**(k+1) )

# The mean and the variance are not defined for all k
mean = (k*l)/(k-1) if k>1 else None
variance = (k*l**2)/(((k-1)**2)*(k-2)) if k>2 else None
std = sqrt(variance) if variance!=None else None
if mean!=None:
    mean_label = rf'$\mathbb{{E}}[X]$'
    plt.axvline(x=mean,color='red',linestyle='--',label=mean_label)
if std!=None and mean!=None: 
    std_label = rf'$\mathbb{{E}}[X]\pm\sqrt{{Var}}[X]$'
    plt.axvspan(mean - std,mean + std,color='red',alpha=0.2,label=std_label)

x_range = np.arange(0,4*l,step)
# Respect discontinuity
y_range = [np.nan if x == l else f(x) for x in x_range] 
plt.title('Pareto distribution\n'+rf'($\lambda={l},\kappa={k}$)')
plt.plot(x_range, y_range)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.show()


```
