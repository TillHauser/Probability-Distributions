# The Uniform Distribution

For real numbers $a$ and $b$ with $a<b$, the probability density function of the *uniform distribution* on the interval $[a,b]$ is given by

$$
f(x):=\begin{cases}
\frac{1}{b-a}, & a\leq x \leq b,\\
0, & \text{otherwise}.
\end{cases}
$$

We write $X\sim\operatorname{Uniform}(a,b)$ to indicate that $X$ is uniformly distributed on $[a,b]$. In the case of $a=0$ and $b=1$, we speak of the *standard uniform distribution*.

**Lemma:** We have

1. $\mathbb{E}[X]=\frac{a+b}{2}$ and
2. $\operatorname{Var}[X]=\frac{(b-a)^2}{12}$.

**Lemma:** 

1. If $X\sim\operatorname{Uniform}(a,b)$, then
$$
\frac{X-a}{b-a}\sim\operatorname{Uniform}(0,1).
$$

2. If $X\sim\operatorname{Uniform}(0,1)$, then
$$
(b-a)X+a\sim\operatorname{Uniform}(a,b).
$$


## Some Visualizations


```python
def Uniform(a=0,b=1): 
    if not a<b:
        raise Exception("a needs to be a number smaller than b for the uniform distrbution.")
    def f(x): 
        if x>=a and x<=b: 
            return 1/(b-a)
        else: 
            return 0
    return f

from matplotlib import pyplot as plt
from math import sqrt
import numpy as np

a = 1
b = 3
f = Uniform(a,b)
x_range = np.arange(-1,5,0.01)



mean = (a+b)/2
std = sqrt((b-a)**2/12)
mean_label = rf'$\mathbb{{E}}[X]$'
plt.axvline(x=mean,color='red',linestyle='--',label=mean_label)
std_label = rf'$\mathbb{{E}}[X]\pm\sqrt{{Var}}[X]$'
plt.axvspan(mean - std,mean + std,color='red',alpha=0.2,label=std_label)

plt.plot(x_range, [f(x) for x in x_range])
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(f"Normal distribution "+rf'($\mu={mu}, \sigma^2={sigma_squared}$)')
plt.grid()
plt.legend()
plt.show()
```

## Simulation

A uniform distribution can easily be simulated with the random module: 

```python
import random

a,b = 2,5
x = random.uniform(a, b)
print(x)

# A whole sample: 
n = 10000
samples = [random.uniform(a, b) for _ in range(n)]
```
