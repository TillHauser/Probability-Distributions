# The Normal Distribution

For $\sigma>0$ and a real number $\mu$ the probability density function of the *normal distribution* is given by 

$$
f(x):=\frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{1}{2\sigma^2}(x-\mu)^2}.
$$
We write $X\sim \operatorname{Normal}(\mu,\sigma^2)$ or $X\sim \operatorname{N}(\mu,\sigma^2)$ to indicate that $X$ is normal distributed. In the case of $\mu=0$ and $\sigma^2=1$ we speak of the *standard normal distribution*. 

**Lemma:** We have 
1) $\mathbb{E}[X]=\mu$ and 
2) $\operatorname{Var}[X]=\sigma^2$. 

**Lemma:** 
1) If $X\sim \operatorname{Normal}(\mu,\sigma^2)$, then $(X-\mu)/\sigma \sim \operatorname{Normal}(0,1)$. 
2) If $X\sim \operatorname{Normal}(0,1)$, then $\sigma X+\mu \sim \operatorname{Normal}(\mu,\sigma^2)$. 

## Some Visualizations

```python
from matplotlib import pyplot as plt
import numpy as np
from math import pi,exp, sqrt


mu = 2
sigma_squared = 5

mean = mu
std = sqrt(sigma_squared)
mean_label = rf'$\mathbb{{E}}[X]$'
plt.axvline(x=mean,color='red',linestyle='--',label=mean_label)
std_label = rf'$\mathbb{{E}}[X]\pm\sqrt{{Var}}[X]$'
plt.axvspan(mean - std,mean + std,color='red',alpha=0.2,label=std_label)


def normal_distribution(mu=0,sigma_squared=1): 
    return lambda x: 1/sqrt(2*pi*sigma_squared) *exp(-1/(2*sigma_squared)*(x-mu)**2)


x_range = np.arange(-20,20,0.01)

f = normal_distribution(mu=mu, sigma_squared=sigma_squared)

plt.plot(x_range, [f(x) for x in x_range])
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(f"Normal distribution "+rf'($\mu={mu}, \sigma^2={sigma_squared}$)')
plt.grid()
plt.legend()
plt.show()

```

## Simulation

The normal distribution can easily be simulated from `random.gauss`. 

```python
import random

mu = 0
sigma = 1
x = random.gauss(mu, sigma)
print(x)

# We can get samples as follows
n = 10000
samples = [random.gauss(mu, sigma) for _ in range(n)]
```
